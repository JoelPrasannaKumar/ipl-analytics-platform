```python
%sql
CREATE DATABASE IF NOT EXISTS ipl_bronze;
CREATE DATABASE IF NOT EXISTS ipl_silver;
CREATE DATABASE IF NOT EXISTS ipl_gold;
```

```python
from pyspark.sql.functions import current_timestamp, lit,col
```

1. Read Raw Data from Unity Catalog Volume

```python
matches_raw_df=spark.read.format("csv").option("header","true")\
    .option("inferSchema","true")\
    .load("/Volumes/workspace/ipl_bronze/raw_data/matches.csv")

deliveries_raw_df=spark.read.format("csv").option("header","true")\
    .option("inferSchema","true")\
    .load("/Volumes/workspace/ipl_bronze/raw_data/deliveries.csv")    
```

2. Add Mandatory Audit Metadata Columns

```python
matches_bronze_df = matches_raw_df \
    .withColumn("ingestion_ts", current_timestamp()) \
    .withColumn("source_system", lit("Kaggle_Manual_Upload")) \
    .withColumn("source_file", col("_metadata.file_path"))

deliveries_bronze_df = deliveries_raw_df \
    .withColumn("ingestion_ts", current_timestamp()) \
    .withColumn("source_system", lit("Kaggle_Manual_Upload")) \
    .withColumn("source_file", col("_metadata.file_path"))
```

#Bronze Layer

Write to Bronze Layer as Delta Tables (Append Mode)

```python
matches_bronze_df.write.format("delta")\
    .mode("overwrite")\
    .saveAsTable("workspace.ipl_bronze.matches")

deliveries_bronze_df.write.format("delta")\
    .mode("overwrite")\
    .saveAsTable("workspace.ipl_bronze.deliveries")
```

```python
print("Bronze ingestion complete!")

display(spark.table("workspace.ipl_bronze.matches").limit(10))

display(spark.table("workspace.ipl_bronze.deliveries").limit(10))
```

#Silver Layer

1. Read from the Bronze Layer

```python
bronze_matches=spark.table("workspace.ipl_bronze.matches")
bronze_deliveries=spark.table("workspace.ipl_bronze.deliveries")

```

2. Data Quality Rules for Deliveries \
Rule 1: Overs must be between 1 and 20 \
Rule 2: batsman_runs cannot be negative\
Rule 3: Batsman name cannot be NULL

```python
silver_deliveries=bronze_deliveries\
    .filter(col("over").between(1,20))\
    .filter(col("batsman_runs")>=0)\
    .filter(col("batsman").isNotNull())
```

Data Quality Rules for Matches\
Let's filter out abandoned matches where there is no winner


```python
silver_matches=bronze_matches.filter(col("winner").isNotNull())
```

Write Cleaned Data to the Silver Layer (Overwrite mode for idempotency)

```python
silver_matches.write.format("delta").mode("overwrite")\
    .saveAsTable("workspace.ipl_silver.matches")

silver_deliveries.write.format("delta").mode("overwrite")\
    .saveAsTable("workspace.ipl_silver.deliveries")
print("Silver layer cleaned and saved!")

display(spark.table("workspace.ipl_silver.matches").limit(10))
```

#Gold Layer (Business & Analytics)

Create temporary views so we can easily query them with SQL

```python
spark.table("workspace.ipl_silver.matches").createOrReplaceTempView("vw_matches")
spark.table("workspace.ipl_silver.deliveries").createOrReplaceTempView("vw_deliveries")
```

1. fact_match_summary (Match-level stats)

```python
fact_match_summary_df=spark.sql("""
    SELECT 
        m.id AS match_id,
        m.season,
        m.city,
        m.team1,
        m.team2,
        m.toss_winner,
        m.toss_decision,
        m.winner,
        SUM(d.total_runs) AS match_total_runs
    FROM vw_matches m
    LEFT JOIN vw_deliveries d ON m.id=d.match_id
    GROUP BY m.id,m.season,m.city,m.team1,m.team2,m.toss_winner, m.toss_decision,m.winner                   
    """)
fact_match_summary_df.write.format("delta").mode("overwrite").saveAsTable("workspace.ipl_gold.fact_match_summary")
```

```python
display(spark.table("workspace.ipl_gold.fact_match_summary").limit(10))
```

2. fact_player_performance (Player-level metrics: Runs, Balls Faced, Wickets)

```python
fact_player_performance_df=spark.sql("""
    WITH batting_stats AS(
        SELECT
            match_id,
            batsman AS player_name,
            SUM(batsman_runs) AS runs_scored,
            COUNT(ball) AS balls_faced
        FROM vw_deliveries
        GROUP BY match_id, batsman 
    ),
    bowling_stats AS(
        SELECT
            match_id,
            bowler AS player_name,
            COUNT(player_dismissed) AS wickets_taken
        FROM vw_deliveries
        GROUP BY match_id, bowler
    )
    SELECT 
        coalesce(b.match_id,w.match_id ) AS match_id,
        coalesce(b.player_name,w.player_name ) AS player_name,
        coalesce(b.runs_scored,0 ) AS runs_scored,
        coalesce(b.balls_faced,0 ) AS balls_faced,
        coalesce(w.wickets_taken,0 ) AS wickets_taken
    FROM batting_stats b 
    FULL OUTER JOIN bowling_stats w
        ON b.match_id=w.match_id AND b.player_name=w.player_name
    """)
fact_player_performance_df.write.format("delta").mode("overwrite").saveAsTable("workspace.ipl_gold.fact_player_performance")
```

Optimize the Delta Tables for fast BI queries

```python
spark.sql("optimize workspace.ipl_gold.fact_match_summary zorder by (match_id)")
spark.sql("optimize workspace.ipl_gold.fact_player_performance zorder by (match_id,player_name)")
```

```python
print("Gold layer built and optimized successfully")
display(spark.table("workspace.ipl_gold.fact_match_summary").limit(10))
display(spark.table("workspace.ipl_gold.fact_player_performance").limit(10))
```
